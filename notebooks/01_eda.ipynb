## 📌 Key Insight: Class Imbalance & Evaluation Metric

The dataset is **highly imbalanced**:
* Class **no** (active customers): **85.86%**
* Class **yes** (churned customers - Positive Class): **14.14%**

### Why F1-Score instead of Accuracy?
If we implement a naive model that predicts "no" (no churn) for every single subscriber, it will achieve an accuracy of **85.86%**. However, it will fail to identify a single churned customer, resulting in a recall of `0.0` for the positive class.

To solve this:
1. We evaluate using **F1-Score** (harmonic mean of Precision and Recall) with `positive class = yes`.
2. We use **Stratified split** in Repeated K-Fold cross-validation to maintain consistent class proportions in every fold.
